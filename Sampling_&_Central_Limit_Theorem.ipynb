```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
np.random.seed(42)
population = pd.Series(np.random.exponential(scale=2.0, size=10000))
```

```python
sample_size = 30
num_samples = 1000
sample_means = [population.sample(sample_size).mean()
    for _ in range(num_samples)]
```

```python
print("Population Mean:", population.mean())
print("Mean of Sample Means:", np.mean(sample_means))
print("Population Std Dev:", population.std())
print("Std Dev of Sample Means:", np.std(sample_means))
print("Standard Error (sigma/sqrt(n)):", population.std() / np.sqrt(sample_size))
```

```python
fig, ax = plt.subplots(1, 2, figsize=(10, 4))
ax[0].hist(population, bins=40,)
ax[0].set_title("Population Distribution")
ax[1].hist(sample_means, bins=30)
ax[1].set_title("Distribution of Sample Means (n = 30)")
plt.tight_layout()
plt.show()
```
