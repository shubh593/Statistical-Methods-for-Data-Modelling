```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
from sklearn.datasets import load_diabetes
```

```python
df = load_diabetes(as_frame=True).frame
cols = ["bmi", "bp", "s5", "target"]
```

```python
pearson = df[cols].corr(method="pearson")
print(pearson)
```

```python
spearman = df[cols].corr(method="spearman")
print(spearman)
```

```python
r, p_r = stats.pearsonr(df["bmi"], df["target"])
print("Pearson r:", r, "p-value:", p_r)
```

```python
rho, p_s = stats.spearmanr(df["bmi"], df["target"])
print("Spearman rho:", rho, "p-value:", p_s)
```

```python
sns.heatmap(pearson, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation Heatmap (Pearson)")
plt.show()
```
