```python
import pandas as pd
from scipy.stats import chi2_contingency
from sklearn.datasets import load_iris
```

```python
df = load_iris(as_frame=True).frame
```

```python
df.head()
```

```python

df["Size"] = pd.qcut(df["sepal length (cm)"], q=3,
                    labels=["Small", "Medium", "Large"])

df["Species"] = df["target"].map({0: "Setosa", 1: "Versicolor", 2: "Virginica"})

table = pd.crosstab(df["Size"], df["Species"])
print(table)
```

```python
chi2, p, dof, expected = chi2_contingency(table)
```

```python
print("Chi-square statistic:", chi2)
print("Degrees of freedom:", dof)
print("p-value:", p)
```

```python
print(pd.DataFrame(expected, index=table.index, columns=table.columns))
```

```python
if p < 0.05:
    print("Reject H0: the variables are dependent")
else:
    print("Fail to reject H0: the variables are independent")
```
