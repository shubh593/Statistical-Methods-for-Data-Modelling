```python
import pandas as pd
from scipy import stats
from sklearn.datasets import load_wine
```

```python
df = load_wine(as_frame=True).frame
alpha = 0.05
```

```python
df.info()
```

```python
df.head()
```

```python
sample = df["alcohol"]
t1, p1 = stats.ttest_1samp(sample, popmean=13.0)

print("--- ONE-SAMPLE T-TEST ---")
print(f"Sample Mean = {sample.mean():.4f}")
print(f"t-statistic = {t1:.4f}, p-value = {p1:.4e}")
```

```python
# Two-sample t-test (H0: mean petal length of class 0 = class 1)
group1 = df[df["target"] == 0]["petal length (cm)"]
group2 = df[df["target"] == 1]["petal length (cm)"]
t2, p2 = stats.ttest_ind(group1, group2, equal_var=False)
print("Two-sample t =", t2, ", p-value =", p2)
```

```python
if p1 < alpha:
  print(f"Decision: p ({p1:.4e}) < {alpha} -> Reject H0")
  print("Interpretation: The mean alcohol content differs significantly from 13.0%.\n")
else:
  print(f"Decision: p ({p1:.4e}) >= {alpha} -> Fail to reject H0")
  print("Interpretation: No significant difference from 13.0%.\n")
```

```python
group1 = df[df["target"] == 0]["alcohol"]
group2 = df[df["target"] == 1]["alcohol"]
t2, p2 = stats.ttest_ind(group1, group2, equal_var=False)
```

```python
print("--- TWO-SAMPLE T-TEST ---")
print(f"Class 0 Mean = {group1.mean():.4f} | Class 1 Mean = {group2.mean():.4f}")
print(f"t-statistic = {t2:.4f}, p-value = {p2:.4e}")
```

```python
if p2 < alpha:
  print(f"Decision: p ({p2:.4e}) < {alpha} -> Reject H0")
  print("Interpretation: Class 0 and Class 1 have significantly different alcohol levels.")
else:
  print(f"Decision: p ({p2:.4e}) >= {alpha} -> Fail to reject H0")
  print("Interpretation: No significant difference between Class 0 and Class 1.")
```
